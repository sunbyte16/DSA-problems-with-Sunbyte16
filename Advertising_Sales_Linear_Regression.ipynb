# Advertising Sales Prediction using Linear Regression

This notebook demonstrates how to predict **sales** using advertising data with **Linear Regression**, without needing an external CSV dataset.

## Problem Name
**Advertising Sales Prediction using Linear Regression**

### Logic
1. Create the advertising dataset.
2. Separate input features (`TV`, `radio`, `newspaper`) and target (`sales`).
3. Split data into training and testing sets.
4. Standardize the input data.
5. Train a Linear Regression model.
6. Predict sales.
7. Calculate Mean Squared Error.

```python
# Import packages
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

## 1. Create the Dataset

```python
# Create dataset directly in Python
df = pd.DataFrame({
    'TV': [230.1, 44.5, 17.2, 151.5, 180.8, 8.7, 57.5, 120.2, 199.8, 66.1],
    'radio': [37.8, 39.3, 45.9, 41.3, 10.8, 48.9, 32.8, 19.6, 2.6, 5.8],
    'newspaper': [69.2, 45.1, 69.3, 58.5, 58.4, 75.0, 23.5, 11.6, 21.2, 24.2],
    'sales': [22.1, 10.4, 12.0, 16.5, 17.9, 7.2, 11.8, 13.2, 9.7, 9.6]
})

df
```

## 2. Check and Clean the Data

```python
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
print('\nNull values:')
print(df.isnull().sum())

# Drop rows containing null values
df.dropna(inplace=True)

print('\nData after removing null values:')
print(df)
```

## 3. Separate Input (X) and Target (y)

```python
X = df.drop('sales', axis=1)
y = df['sales']

print('X:')
print(X)

print('\ny:')
print(y)
```

## 4. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=101
)

print('X_train shape:', X_train.shape)
print('X_test shape:', X_test.shape)
print('y_train shape:', y_train.shape)
print('y_test shape:', y_test.shape)
```

## 5. Standardize the Data

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print('Training data standardized successfully.')
```

## 6. Train the Linear Regression Model

```python
model = LinearRegression()
model.fit(X_train, y_train)

print('Model trained successfully.')
```

## 7. Predict Sales

```python
y_pred = model.predict(X_test)

print('Predicted Sales:')
print(y_pred)
```

## 8. Calculate Mean Squared Error

```python
mse = mean_squared_error(y_test, y_pred)

print('Mean Squared Error:', mse)
```

## Conclusion

The Linear Regression model was trained using advertising data and used to predict sales. The **Mean Squared Error (MSE)** measures the average squared difference between the actual and predicted sales values.
